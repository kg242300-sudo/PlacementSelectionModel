```python
df = pd.DataFrame(data)
```

```python
#Creating a dataframe
df
```

```python
#select the features
# Roll number and Name are nor used for predection
X = df[
    ["CGPA",
     "Skills",
     "Projects",
     "Internships",
     "Certifications",
     "DSA_Score"
     ]
]

```

```python
X
```

```python
# slect the target /label (y)
y = df["Placed"]
```

```python
y
```

```python
#6 splitting the data into train and test data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2, # 20% test data
    random_state=42 # Random number
)
```

```python
print("Training students:",len(X_train))
print("Testing students:",len(X_test))
```

Creating the model for the project

```python
model = DecisionTreeClassifier(random_state=42)
```

Train the model

```python
model.fit(X_train, y_train)
print('\nModel Trained Successfully')
```

Test the model

```python
Guessing = model.predict(X_test)
print('\nActual Values', list(y_test))
print('\nPredicted Values', list(Guessing))
```

Testing Accuracy

```python
accuracy = accuracy_score(y_test, Guessing)
print('\nAccuracyscore', accuracy)
```

```python
from sklearn.metrics import classification_report

# Generate and print the classification report
report = classification_report(y_test, Guessing, target_names=['Not Placed', 'Placed'])
print("Classification Report:\n")
print(report)
```
